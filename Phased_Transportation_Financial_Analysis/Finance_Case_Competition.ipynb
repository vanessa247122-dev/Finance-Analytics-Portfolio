```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

# ── Data ──────────────────────────────────────────────────
options = ['Transit / Rideshare', 'Hybrid\n(Transit → Lease)', 'Buy']
costs   = [22800.00, 36143.62, 38419.35]
colors  = ['#3DBFAA', '#A78BFA', '#F4A460']

# ── Figure setup ──────────────────────────────────────────
BG        = '#0D1130'   # outer slide background (navy)
CHART_BG  = '#3A3A3A'   # chart area gray (matching your deck)

fig, ax = plt.subplots(figsize=(14, 6))
fig.patch.set_facecolor(BG)
ax.set_facecolor(CHART_BG)

# ── Bars ──────────────────────────────────────────────────
bars = ax.barh(options, costs, color=colors, height=0.55, zorder=2)

# ── Value labels ──────────────────────────────────────────
for bar, cost in zip(bars, costs):
    x_pos = bar.get_width()
    if x_pos > 8000:
        ax.text(x_pos - 1000, bar.get_y() + bar.get_height() / 2,
                f'${cost:,.2f}',
                va='center', ha='right',
                color='white', fontsize=13, fontweight='bold')
    else:
        ax.text(x_pos + 500, bar.get_y() + bar.get_height() / 2,
                f'${cost:,.2f}',
                va='center', ha='left',
                color='white', fontsize=13, fontweight='bold')

# ── Axes formatting ───────────────────────────────────────
ax.set_xlim(0, 50000)
ax.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f'${int(x):,}'))
ax.tick_params(axis='x', colors='white', labelsize=11)
ax.tick_params(axis='y', colors='white', labelsize=13)
ax.set_xlabel('Total 5-Year Cost', color='white', fontsize=12, labelpad=10)
ax.yaxis.label.set_color('white')
ax.xaxis.grid(True, color='#555555', linestyle='--', linewidth=0.8, zorder=1)
ax.set_axisbelow(True)

# ── Spines ────────────────────────────────────────────────
for spine in ax.spines.values():
    spine.set_visible(False)

# ── Annotation ────────────────────────────────────────────
ax.annotate(
    '+$13,344 vs Transit\n(cost of flexibility)',
    xy=(36143.62, 1),
    xytext=(36143.62, 1.45),
    color='#A78BFA', fontsize=9, ha='center',
    arrowprops=dict(arrowstyle='->', color='#A78BFA', lw=1.2)
)

# ── Title ─────────────────────────────────────────────────
ax.set_title("Jordan's 5-Year Transportation Cost Comparison",
             color='white', fontsize=16, fontweight='bold', pad=18)

plt.tight_layout()
plt.savefig('slide3_chart.png', dpi=200,
            bbox_inches='tight', facecolor=BG)
plt.show()
print("Saved → slide3_chart.png")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ── Data ──────────────────────────────────────────────────
options = ['Buy', 'Hybrid\n(Blended Avg)', 'Transit / Rideshare']
costs   = [781.67, 602.39, 380.00]
colors  = ['#F4A460', '#A78BFA', '#3DBFAA']

# ── Figure setup ──────────────────────────────────────────
BG       = '#0D1130'
CHART_BG = '#3A3A3A'

fig, ax = plt.subplots(figsize=(10, 6))
fig.patch.set_facecolor(BG)
ax.set_facecolor(CHART_BG)

# ── Bars ──────────────────────────────────────────────────
bars = ax.bar(options, costs, color=colors, width=0.5, zorder=2)

# ── Value labels on top of bars ───────────────────────────
for bar, cost in zip(bars, costs):
    ax.text(bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 8,
            f'${cost:,.2f}',
            ha='center', va='bottom',
            color='white', fontsize=13, fontweight='bold')

# ── Axes formatting ───────────────────────────────────────
ax.set_ylim(0, 950)
ax.yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f'${int(x):,}'))
ax.tick_params(axis='y', colors='white', labelsize=11)
ax.tick_params(axis='x', colors='white', labelsize=12)
ax.set_ylabel('Monthly Cost', color='white', fontsize=12, labelpad=10)
ax.yaxis.grid(True, color='#555555', linestyle='--',
              linewidth=0.8, zorder=1)
ax.set_axisbelow(True)

# ── Spines ────────────────────────────────────────────────
for spine in ax.spines.values():
    spine.set_visible(False)

# ── Hybrid footnote ───────────────────────────────────────
ax.text(1, 602.39 + 45,
        '* blended avg\nacross both phases',
        ha='center', va='bottom',
        color='#A78BFA', fontsize=8, style='italic')

# ── Title ─────────────────────────────────────────────────
ax.set_title('Monthly Transportation Burden by Option',
             color='white', fontsize=15,
             fontweight='bold', pad=18)

plt.tight_layout()
plt.savefig('slide7_chart.png', dpi=200,
            bbox_inches='tight', facecolor=BG)
plt.show()
print("Saved → slide7_chart.png")
```
